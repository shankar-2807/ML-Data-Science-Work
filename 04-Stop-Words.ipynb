# Stop Words
Words like "a" and "the" appear so frequently that they don't require tagging as thoroughly as nouns, verbs and modifiers. We call these *stop words*, and they can be filtered from the text to be processed. spaCy holds a built-in list of some 326 English stop words.

```python
# Perform standard imports:
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
# Print the set of spaCy's default stop words (remember that sets are unordered):
print(nlp.Defaults.stop_words)
```

```python
len(nlp.Defaults.stop_words)
```

### Remove word from stop words

```python
nlp.Defaults.stop_words.remove("never")
# print(nlp.Defaults.stop_words)
nlp.vocab["never"].is_stop=False
# nlp.vocab
```

```python
len(nlp.Defaults.stop_words)
```

### Add word from stop words

```python
nlp.vocab['product'].is_stop=True
nlp.Defaults.stop_words.add('product')
```

```python
len(nlp.Defaults.stop_words)
```

### Remove multiple words

```python
words = ["is","am"]
for w in words:
    nlp.Defaults.stop_words.remove(w)
    nlp.vocab[w].is_stop=False
```

```python
len(nlp.Defaults.stop_words)
```
