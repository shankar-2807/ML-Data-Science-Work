```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("titanic data.csv")
```

```python
dataset
```

```python
dataset.info()
```

```python
dataset = dataset.drop('Cabin',axis=1)
```

```python
dataset.info()
```

```python
dataset.loc[dataset["Age"].isna(),"Age"] = dataset["Age"].mean()
```

```python
dataset.info()
```

```python
dataset[dataset['Embarked'].isna()] 
```

```python
dataset.dropna(inplace=True)
```

```python
dataset.info()
```

```python
plt.boxplot(x=dataset['Age'])
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
dataset["Sex"]=le.fit_transform(dataset["Sex"]) 
```

```python
le1 = LabelEncoder()
dataset["Embarked"]=le1.fit_transform(dataset["Embarked"]) 
```

```python
X = dataset[["Pclass", "Age","Sex", "SibSp", "Parch", "Embarked"]]
```

```python
y= dataset["Survived"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=30)
```

```python
import tensorflow as tf 
```

```python
model = tf.keras.models.Sequential()
model.add(tf.keras.layers.Dense(units = 6,activation="softmax"))
model.add(tf.keras.layers.Dense(units = 4,activation="softmax"))
model.add(tf.keras.layers.Dense(units = 1,activation="sigmoid"))
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
model.fit(X_train,y_train,batch_size=32,epochs=5)
```

```python
import shap
explainer = shap.Explainer(model,X_train)
shap_values = explainer(X_test)
shap.plots.bar(shap_values)
```
